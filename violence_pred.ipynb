```python
import cv2
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import load_model

# Load the trained model, specifying custom_objects if necessary
# For example, if you used a custom activation function:
# custom_objects = {'custom_activation': your_custom_activation_function}
# model = load_model("modelnew.h5", custom_objects=custom_objects)
model = load_model("modelnew.h5")

# Define video input
video_path = "V_19.mp4"  # Change this to your video file
cap = cv2.VideoCapture(video_path)

# Check if the video opened successfully
if not cap.isOpened():
    print("Error: Could not open video.")
    exit()

# Define frame processing function
def preprocess_frame(frame):
    # Resize to match the input size of your model
    # This should be the same size as IMG_SIZE used during training
    frame = cv2.resize(frame, (128, 128))
    frame = frame / 255.0  # Normalize
    frame = np.expand_dims(frame, axis=0)  # Add batch dimension
    return frame

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break  # Exit loop if video ends

    # Preprocess frame
    processed_frame = preprocess_frame(frame)

    
    # Predict violence probability
    prediction = model.predict(processed_frame)[0][0]  # Assuming binary classification
    label = "Violence" if prediction > 0.5 else "Non-Violence"

    print(label)
   

    # Display result on frame
    color = (0, 0, 255) if label == "Violence" else (0, 255, 0)
    cv2.putText(frame, f"{label}: {prediction:.2f}", (10, 50),
                cv2.FONT_HERSHEY_SIMPLEX, 1, color, 2, cv2.LINE_AA)

    # Show frame
    cv2.imshow("Violence Detection", frame)

    # Press 'q' to exit
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

# Release resources
cap.release()
cv2.destroyAllWindows()
```
